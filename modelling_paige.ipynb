## Loading Datasets:
After running this block, you have loaded all the data from the CSVs into train and test

```python
# 1. Load the datasets

folders = [
    path.parent
    for path in Path("/kaggle/input").rglob("sample_submission.csv")
    if (path.parent / "train.csv").exists()
    and (path.parent / "test.csv").exists()
]

assert len(folders) == 1, f"Expected one folder, found: {folders}"
DATA_DIR = folders[0]

train = pd.read_csv(DATA_DIR / "train.csv")
test = pd.read_csv(DATA_DIR / "test.csv")
sample_submission = pd.read_csv(DATA_DIR / "sample_submission.csv")

print("Loaded from:", DATA_DIR)
print("Training shape:", train.shape)
print("Test shape:", test.shape)

display(train.head())
```

```python
# Explore train dataset
train.info()

display(train.isna().sum().rename("missing_count").to_frame())
display(train["is_fraud"].value_counts())
display(sample_submission.head())
```

```python
# Separate features from the target

feature_columns = [
    col for col in train.columns
    if col not in ["claim_id", "is_fraud"]
]

X = train[feature_columns].copy()
y = train["is_fraud"].copy()
X_test = test[feature_columns].copy()
```

```python
# A validation split

from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42,
)

print("Training rows:", len(X_train))
print("Validation rows:", len(X_val))
```
